```python
import json
```

```python
def load_data(filename):
    with open(filename,"r+") as f:
        data = json.load(f)
    return data
data = load_data("store_data.json")
print(data)
```

```python
def clean_data(data):
    text_to_num = {"one":1,"two":2,"three":3,"four":4,"five":5}
    unique_users = set()
    cleaned_data = []
    for user in data:
        #clean rating
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating
        
    for user in data:
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

    for user in data:
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"].strip())
        cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
```

```python
with open("store_data.json","w") as f:
    json.dump(data,f,indent = 4,sort_keys = True)
```

```python
with open("store_data.json","r") as f:
    print(json.load(f))
```

```python
#Get meaningful insights from the data


def get_insights(data):
    total_users = 0
    sum_of_ratings = 0
    low_ratings_sum = 0
    users_with_low_rating = 0
    
    for user in data:
        total_users += 1
        rating = float(user.get("rating"))
        sum_of_ratings += rating
        
        if rating < 3:
            users_with_low_rating += 1
    print(f"{(users_with_low_rating/total_users) * 100}% of users gave low rating.")
    print(f"{sum_of_ratings/total_users} is avg rating of the total users.")
        
```

```python
get_insights(data)
```
