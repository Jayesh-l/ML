```python
import pandas as pd

#reading with pandas (iris data.... 'IRIS.csv')
iris_original = pd.read_csv('IRIS.csv')
iris_original
```

```python
iris_copy = iris_original.copy()
iris_copy
```

```python
#showing only first ten rows
iris_copy.head(10)

#showing datatypes and shape of the data
iris_copy.dtypes
iris_copy.shape

#summary statistics (mean,std,min,max)
iris_copy.describe()

#rows with petal length > 4.5
iris_copy[iris_copy["petal_length"]>4.5]

#species="Iris-virginica"
iris_copy[iris_copy["species"] == "Iris-virginica"]

#Group by species and compute
#all the statistics of the "sepal_length","petal_width","sepal_width"
iris_copy[["sepal_length","petal_width","sepal_width"]].describe()
```

```python
#using aggregate fuction to calculate...
#Average sepal_length
#Maximum petal_width
#Standard deviation of sepal_width
iris_copy.agg({
    "sepal_length" : "mean",
    "petal_width" : "max",
    "sepal_width" : "std",
})
```

```python
#using groupby function to group data by...
#Average sepal_length
#Maximum petal_width
#Standard deviation of sepal_width

iris_copy.groupby("sepal_length")["petal_length"].mean()
#left column is the sepal_length and the right column is the avg of the petal length of the particular sepal_length

# here i grouped by species and calculated the mean/avg of the sepal_length
iris_copy.groupby("species")["sepal_length"].mean()

#here i goruped by species and calcualte the max of the petal width for the particular species
iris_copy.groupby("species")["petal_width"].max()

#if you read above comments you will now understand the below line
iris_copy.groupby("species")["sepal_width"].std()
```

```python
#now im creating a new column
#finding the avg petal_ratio
iris_copy["petal_ratio"] = iris_copy["petal_length"]/iris_copy["petal_width"]
iris_copy
```
