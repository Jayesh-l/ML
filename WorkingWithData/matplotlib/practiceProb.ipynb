```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
days = ["Mon","Tue","Wed","Thu","Fri"]
cities = ["New York","London","Delhi","Tokyo"]
temperatures = [
    [22,23,21,24,25],
    [18,19,17,20,21],
    [30,32,31,33,34],
    [25,26,24,27,28]
]
fig,ax = plt.subplots(2,2)
_ = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days,temperatures[_],marker = 'o')
        ax[i][j].set_title(cities[_])
        ax[i][j].grid(True)
        _ = _+1;
fig.suptitle("Temperature of Cities over the week")
fig.supylabel("Temperature in Celsius")
fig.supxlabel("Days")
fig.tight_layout()
```
